```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_digits
import warnings
warnings.filterwarnings('ignore')
```

```python
digits=load_digits()
df=pd.DataFrame(digits.data,columns=digits.feature_names)
df['target']=digits.target
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().any().sum()
```

```python
df.duplicated().any().sum()
```

```python
X=df.drop('target',axis='columns')
X.head()
```

```python
Y=df.target
Y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.decomposition import PCA
pca=PCA(0.95)
X_train=pca.fit_transform(X_train)
X_test=pca.transform(X_test)
```

```python
pca.n_components_
```

```python
from sklearn.preprocessing import MinMaxScaler
mms=MinMaxScaler()
X_train=mms.fit_transform(X_train)
X_test=mms.transform(X_test)
```

It is multiclass classification problem

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
logistic_model=LogisticRegression()
logistic_model.fit(X_train,Y_train)
logistic_test_score=accuracy_score(Y_test,logistic_model.predict(X_test))
logistic_train_score=accuracy_score(Y_train,logistic_model.predict(X_train))
print("Test Score: ",logistic_test_score)
print("Train Score: ",logistic_train_score)
```

```python
from sklearn.tree import DecisionTreeClassifier
decision_tree_model=DecisionTreeClassifier()
decision_tree_model.fit(X_train,Y_train)
decision_tree_test_score=accuracy_score(Y_test,decision_tree_model.predict(X_test))
decision_tree_train_score=accuracy_score(Y_train,decision_tree_model.predict(X_train))
print("Test Score: ",decision_tree_test_score)
print("Train Score: ",decision_tree_train_score)
```

```python
from sklearn.ensemble import RandomForestClassifier
random_forest_model=RandomForestClassifier(n_estimators=201)
random_forest_model.fit(X_train,Y_train)
random_forest_test_score=accuracy_score(Y_test,random_forest_model.predict(X_test))
random_forest_train_score=accuracy_score(Y_train,random_forest_model.predict(X_train))
print("Test Score: ",random_forest_test_score)
print("Train Score: ",random_forest_train_score)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=11)
knn_model.fit(X_train,Y_train)
knn_model_test_score=accuracy_score(Y_test,knn_model.predict(X_test))
knn_model_train_score=accuracy_score(Y_train,knn_model.predict(X_train))
print("Test Score: ",knn_model_test_score)
print("Train Score: ",knn_model_train_score)
```

```python
from sklearn.svm import SVC
svc_model=SVC()
svc_model.fit(X_train,Y_train)
svc_model_test_score=accuracy_score(Y_test,svc_model.predict(X_test))
svc_model_train_score=accuracy_score(Y_train,svc_model.predict(X_train))
print("Test Score: ",svc_model_test_score)
print("Train Score: ",svc_model_train_score)
```

```python
from sklearn.naive_bayes import GaussianNB
gaussian_nb_model=GaussianNB()
gaussian_nb_model.fit(X_train,Y_train)
gaussian_nb_model_test_score=accuracy_score(Y_test,gaussian_nb_model.predict(X_test))
gaussian_nb_model_train_score=accuracy_score(Y_train,gaussian_nb_model.predict(X_train))
print("Test Score: ",gaussian_nb_model_test_score)
print("Train Score: ",gaussian_nb_model_train_score)
```

Best is SVM

```python
import cv2
def predict(img):
    if img is None:
        return "No input"
    img_resized = cv2.resize(img, (8, 8), interpolation=cv2.INTER_AREA)
    img_rescaled = (img_resized / 255.0) * 16.0
    img_flattened = img_rescaled.reshape(1, -1)
    img_pca = pca.transform(img_flattened)
    img_scaled = mms.transform(img_pca)
    prediction = svc_model.predict(img_scaled)
    return int(prediction[0])
```

```python
import gradio as gr
ifc = gr.Interface(
    fn=predict,
    inputs=gr.Sketchpad(
        type="numpy",
        image_mode='L',
        shape=(28, 28)
    ),
    outputs="text",
    title="SVC Model Prediction"
)

# Launch the app
ifc.launch(share=True)
```
